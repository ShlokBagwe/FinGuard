```python
df = pd.read_parquet('/kaggle/input/datasets/shlokbagwe/main-classification-df-v2/classification_df_v2.parquet')
df.head()
```

```python
df = df.sort_values(by=['transaction_datetime']).reset_index(drop=True)
```

```python
df.shape
```

```python
split_point = int(len(df) * 0.8)
split_date = df.iloc[split_point]['transaction_datetime']

train_2 = df[df['transaction_datetime'] < split_date].copy()
test_2 = df[df['transaction_datetime'] >= split_date].copy()
```

```python
cols = ['client_id_x', 'current_age', 'mcc', 'merchant_city','merchant_state', 'transaction_datetime', 'date', 'time']
train_df = pd.concat(
    [train_df, train_2[cols]],
    axis=1
)
test_df = pd.concat([test_df, test_2[cols]], axis=1)
```

```python
train_2.groupby('mcc')['target_numeric'].mean().sort_values(ascending=False)
```

```python
train_df.drop(columns = ['gender', 'card_brand', 'has_chip', 'debt_to_income', 'num_cards_issued'], inplace=True, axis=1)
```

```python
test_df.drop(columns = ['gender', 'card_brand', 'has_chip', 'debt_to_income', 'num_cards_issued'], inplace=True, axis=1)
```

```python
train_df.columns
```

```python
train_df['mcc'].value_counts()
```

```python
train_df['merch_total_txns'] = train_df.groupby('mcc').cumcount()
test_df['merch_total_txns'] = test_df.groupby('mcc').cumcount()
```

```python
train_df['prev_mcc_txns'] = train_df.groupby('mcc').cumcount()

temp = train_df.groupby('mcc')['target_numeric'].cumsum()

train_df['prev_mcc_frauds'] = temp.groupby(train_df['mcc']).shift(1).fillna(0)

train_df['previous_mcc_risk_score'] = (
    train_df['prev_mcc_frauds'] / train_df['prev_mcc_txns']
)


train_df['previous_mcc_risk_score'] = (
    train_df['previous_mcc_risk_score']
      .replace([np.inf, -np.inf], 0)
      .fillna(0)
)
```

```python
# Creating previous mcc risk score for diff types of mccs

test_df['prev_mcc_txns'] = test_df.groupby('mcc').cumcount()

temp = test_df.groupby('mcc')['target_numeric'].cumsum()

test_df['prev_mcc_frauds'] = temp.groupby(test_df['mcc']).shift(1).fillna(0)

test_df['previous_mcc_risk_score'] = (
    test_df['prev_mcc_frauds'] / test_df['prev_mcc_txns']
)

test_df['previous_mcc_risk_score'] = (
    test_df['previous_mcc_risk_score']
      .replace([np.inf, -np.inf], 0)
      .fillna(0)
)
```

```python
train_df.drop(columns = ['prev_mcc_txns', 'prev_mcc_frauds'], inplace=True)
test_df.drop(columns = ['prev_mcc_txns', 'prev_mcc_frauds'], inplace=True)
```

```python
top_10_states = train_df['merchant_state'].value_counts().head(10)
```

```python
train_df['merchant_state'] = train_df['merchant_state'].apply(lambda x: x if x in top_10_states or x == "online" else "Other_State")
```

```python
top_10_cities = (
    train_df.groupby('merchant_city').agg(
          total_transactions=('target_numeric', 'size'),
          fraud_transactions=('target_numeric', 'sum'),
          fraud_rate=('target_numeric', 'mean')
      )
      .sort_values('total_transactions', ascending=False)
      .head(10)
      .reset_index()
)

top_10_cities['fraud_rate'] *= 100 
top_10_cities
```

```python
top_10_states = (
    train_df.groupby('merchant_state').agg(
          total_transactions=('target_numeric', 'size'),
          fraud_transactions=('target_numeric', 'sum'),
          fraud_rate=('target_numeric', 'mean')
      )
      .sort_values('total_transactions', ascending=False).head(10)
      .reset_index()
)

top_10_states['fraud_rate'] *= 100 
top_10_states
```
