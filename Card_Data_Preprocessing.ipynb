### has_chip

```python
df['has_chip'] = df['has_chip'].map({'YES': True, 'NO': False})
df['has_chip'].value_counts()
```

### credit_limit

```python
df['credit_limit'] = df['credit_limit'].str.replace("$", "").str.strip().astype(float)
```

### acc_open_date

```python
df['acct_open_date'] = pd.to_datetime(df['acct_open_date'], format = '%m/%Y').dt.to_period('M')
```
